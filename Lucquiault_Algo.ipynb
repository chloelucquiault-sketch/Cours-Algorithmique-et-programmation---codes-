```python
def somme_ite(n):
    somme = 0
    for i in range(1,n+1):
        somme = somme + i 
    return somme

n=3

print("La somme des",n,"premiers entiers est", somme_ite(n))
```

```python
def somme_b2_1(x,y):
    sommeb10 = x + y
    quotient = sommeb10 // 2
    reste = sommeb10 - (quotient * 2)
    return reste

print("Le résultat est ",somme_b2_1(3,7))

def somme_b2_2(x,y):
    if (x+y) % 2 == 0 : #somme paire
        return 0
    if (x+y) % 2 == 1 : #somme impaire
        return 1


print("Le résultat est ",somme_b2_2(3,7))
    
```
